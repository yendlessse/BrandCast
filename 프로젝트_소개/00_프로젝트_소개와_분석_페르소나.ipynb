## 1. 프로젝트 한눈에 보기

BrandCast는 방송콘텐츠의 단순 인기순위를 만드는 프로젝트가 아니다. **핵심 고객층, 온라인 관심, 소비자 반응, 콘텐츠 맥락, 관심의 지속성, 브랜드 위험**을 차례로 검토하여 뷰티 브랜드가 우선적으로 살펴볼 광고 매체 후보와 그 이유를 설명하는 의사결정 분석이다.

| 항목 | 확정 설정 |
|---|---|
| 프로젝트 목표 | 신규 스킨케어 제품의 인지도 확대를 위한 방송콘텐츠 광고 매체 후보 검토 |
| 분석 사용자 | 뷰티 브랜드 마케팅팀 / 미디어 플래너 |
| 핵심 고객 | 20대 및 30대 여성 |
| 브랜드 가치 | 건강한 아름다움, 피부관리, 자기관리 |
| 캠페인 방식 | 방송 PPL·제품 협찬과 SNS 연계 캠페인 |
| 비교 범위 | 예능·드라마를 포함한 전체 유효 방송콘텐츠 |
| 분석 기간 | 최신 완료 기준일로부터 이용 가능한 최근 1년 |

## 2. 왜 새로운 선정 기준이 필요한가

기업의 마케팅 담당자는 한정된 자원으로 브랜드 인지도를 높이고 핵심 고객과의 접점을 확대해야 한다. 방송 광고와 PPL은 강력한 접점을 만들 수 있지만, 프로그램의 시청 규모나 화제성이 높다는 사실만으로 특정 뷰티 브랜드에 적합하다고 판단하기는 어렵다.

| 기존 평가 방식 | 놓칠 수 있는 문제 | BrandCast의 보완 질문 |
|---|---|---|
| 전체 시청 규모 중심 | 핵심 고객층이 실제로 얼마나 포함됐는지 알기 어렵다 | 20~30대 여성의 집중도와 도달 규모는 충분한가? |
| 화제성 순위 중심 | 규모에 비해 기대 이상의 관심을 만드는 콘텐츠를 놓칠 수 있다 | 시청 규모를 고려해도 온라인 관심이 높은가? |
| 미디어 노출 중심 | 소비자의 직접적인 반응이 활발한지 구분하기 어렵다 | 미디어 관심보다 시청자 반응이 두드러지는가? |
| 프로그램 이미지 중심 | 제작진이 제시하는 이미지와 소비자 관심 주제가 다를 수 있다 | 실제 관심 키워드가 제품 메시지와 연결되는가? |
| 현재 인기도 중심 | 관심이 상승 중인지 하락 중인지 반영하기 어렵다 | 지금 관심이 성장하거나 안정적으로 유지되는가? |
| 전체 반응량 중심 | 논란이나 부정 이슈가 만든 관심을 구분하기 어렵다 | 브랜드 이미지에 부담이 되는 위험 신호가 있는가? |

따라서 BrandCast는 여러 지표를 한 번에 섞어 단일 순위를 만드는 대신, 마케팅 담당자의 실제 검토 순서에 맞춰 질문을 하나씩 해결한다.

## 4. 왜 RACOI 데이터를 사용하는가

RACOI는 방송콘텐츠의 시청 규모만이 아니라 인터넷반응, 종합반응, 시청층, 키워드와 감성 관련 정보를 함께 검토할 수 있어 후보를 여러 관점에서 비교하는 출발점이 된다. 각 자료의 행 단위와 제공 범위는 실제 수집 단계에서 확인하며, 제공되지 않는 값을 임의로 만들지 않는다.

| 데이터 원천 | 확인하려는 내용 | 프로젝트에서의 역할 | 주의점 |
|---|---|---|---|
| 인터넷반응 주간 자료 | 게시글·댓글·조회·뉴스·영상 등 주간 반응 | Q1·Q3·Q5의 주간 비교와 추세 | 주간 단위와 누락 주차를 구분 |
| 종합반응 월간 자료 | 시청 규모와 온라인 종합반응 | Q2의 시청 규모 대비 관심 비교 | 주간 자료와 직접 혼합하지 않고 월 단위로 집계 |
| 연령·성별 시청층 | 타깃 시청 비중과 규모 | Q1의 20~30대 여성 접점 평가 | 비중과 시청자 수의 모집단 일치 여부 확인 |
| 댓글반응 상세·키워드 | 핵심 관심어, 연관어, 감성 정보 | Q4 콘텐츠 맥락과 Q6 위험 검토 | 원문 댓글보다 공개된 집계 결과를 우선 |

### 분석 범위 원칙

- 최신 완료 기준일로부터 이용 가능한 **최근 1년**을 전체 검토 범위로 삼는다.
- Q1~Q6의 판단 기간은 각 질문의 목적과 실제 데이터 커버리지에 맞춰 고정하고 기록한다.
- 현재 광고 집행 가능 후보와 종영작을 포함한 역사 비교집단을 구분한다.
- 미수집, 결측, 실제 0과 분석 제외를 서로 다른 상태로 관리한다.
- 모든 전처리·집계·지표·후보 순위는 MariaDB SQL에서 계산한다.

### 질문이 이어지는 논리

1. **Target Audience** — 먼저 우리 고객이 존재하는 후보를 확보한다.
2. **Attention Performance** — 그중 시청 규모 이상의 관심을 만드는 후보를 찾는다.
3. **Consumer Response** — 관심이 소비자의 직접 반응으로 나타나는지 확인한다.
4. **Contextual Fit** — 제품 메시지를 자연스럽게 연결할 관심 주제를 찾는다.
5. **Momentum & Persistence** — 지금 집행을 검토할 만한 성장·지속성을 확인한다.
6. **Brand Risk** — 부정 이슈와 브랜드 위험을 검토한 뒤 최종 판단한다.

## 7. 질문별 시각화 계획

소개 노트북에서는 프로젝트를 설명하는 표만 사용한다. 실제 분석 그래프는 Q1~Q6 노트북에서 질문마다 정확히 두 개씩 만들며, 첫 번째 그래프는 현상과 관계를 이해하고 두 번째 그래프는 후보를 비교하는 역할을 맡는다.

| 연구 질문 | 현상 이해 그래프 | 후보 비교 그래프 |
|---|---|---|
| Q1 타깃 고객층 | 연령·성별 Lift 히트맵 | 타깃 시청 규모 순위 막대그래프 |
| Q2 관심 창출력 | 시청 규모·온라인 관심 회귀 산점도 | 기대 관심 대비 초과 성과 순위 |
| Q3 소비자 반응 | 미디어·시청자 버즈 4분면 | 버즈 격차 덤벨 차트 |
| Q4 콘텐츠 맥락 | 핵심 키워드 산점도 | 관심 주제 히트맵 |
| Q5 성장·지속성 | 최근 8주 화제성 추이 | 현재 화제성·성장률 산점도 |
| Q6 브랜드 위험 | 긍정·중립·부정 감성 분포 | 부정 감성 변화 추이 |

> 그래프는 SQL 결과만 입력으로 사용하며 기간, 단위, 표본 수와 출처를 함께 표시한다.

## 8. 분석 결과가 지원하는 마케팅 의사결정

| 의사결정 | 답하려는 내용 | 연결 질문 |
|---|---|---|
| **WHO** 누구에게 광고할 것인가? | 20~30대 여성 핵심 고객층 | Q1 |
| **WHERE** 어떤 프로그램에 광고할 것인가? | 타깃 고객층과 온라인 관심을 확보한 콘텐츠 | Q1·Q2·Q3 |
| **HOW** 어떤 방식으로 연결할 것인가? | 방송 PPL과 소비자 반응을 활용한 디지털 연계 | Q3 |
| **WHAT** 어떤 메시지를 전달할 것인가? | 관심 주제와 연결되는 뷰티·피부관리·자기관리 메시지 | Q4 |
| **WHEN** 언제 집행을 검토할 것인가? | 관심이 성장하거나 안정적으로 유지되는 시점 | Q5 |
| **RISK** 무엇을 추가 검토할 것인가? | 부정 반응과 브랜드 이미지 위험 신호 | Q6 |

## 9. 최종 후보를 판단하는 방식

최종 후보는 특정 지표 하나가 가장 높은 프로그램으로 정하지 않는다. 퍼널을 통과한 근거와 전체 유효 프로그램의 종합 평가를 함께 보고, 캠페인의 핵심인 타깃 도달과 콘텐츠 맥락을 더 중요하게 검토한다.

| 평가 요소 | 초기 비중 | 해석 |
|---|---:|---|
| 핵심 고객층 도달 | 35% | 타깃 규모와 시장 대비 집중도 |
| 콘텐츠 맥락 적합성 | 25% | 뷰티·피부관리·자기관리 관심 주제 |
| 온라인 관심 창출력 | 15% | 시청 규모 대비 관심 초과 성과 |
| 소비자 반응 잠재력 | 15% | 미디어 관심 대비 시청자 반응 |
| 성장 및 지속성 | 10% | 현재 관심, 최근 성장과 안정성 |
| 브랜드 위험 | 별도 검토 | 자동 점수 확정이 아닌 REVIEW·HOLD 판단 |

이 비중은 광고 성과로 학습된 값이 아니라 **신제품 인지도 캠페인을 위한 초기 설계안**이다. 가중치를 바꾸어도 결론이 유지되는지 민감도를 확인하고, 필수 데이터가 부족한 후보는 임의의 0점으로 채우지 않는다.

## 10. 최종 산출물

결과에는 프로그램명과 순위만 제시하지 않는다. 각 후보가 선택되거나 제외된 이유, 데이터 충분성, 적용 기간, 기준값과 해석 한계를 함께 남겨 같은 조건에서 다시 확인할 수 있게 한다.

## 11. 기존 분석과의 차별성

| 일반적인 방송콘텐츠 분석 | BrandCast |
|---|---|
| 어떤 프로그램이 인기 있는가? | 우리 브랜드에 어떤 프로그램이 적합한가? |
| 전체 시청자 규모 비교 | 핵심 고객층 중심의 매체 평가 |
| 온라인 화제성 순위 | 시청 규모 대비 기대 관심의 초과 성과 |
| 전체 미디어 반응량 | 미디어 관심과 소비자 반응의 차이 |
| 프로그램 장르·이미지 비교 | 소비자 관심 주제와 브랜드 맥락 비교 |
| 현재 인기도 평가 | 관심의 성장 방향과 지속성 평가 |
| 개별 분석 결과 나열 | 단계별 후보 선별과 하나의 최종 의사결정 |

BrandCast의 차별점은 분석 자체가 목적이 아니라, 구체적인 마케팅 페르소나의 문제에서 출발해 모든 질문을 **설명 가능한 광고 매체 선정 과정**으로 연결한다는 데 있다.

## 12. 분석 한계와 올바른 해석

<div style="padding:24px 28px; border-radius:16px; background:#fffbeb; border:1px solid #fde68a; color:#78350f;">
  <h3 style="margin:0 0 10px; color:#92400e;">결과는 ‘광고 효과의 보장’이 아니라 ‘우선 검토 근거’다.</h3>
  <p style="margin:0; line-height:1.7;">RACOI 반응 자료는 방송콘텐츠의 상대적인 마케팅 잠재력을 비교하는 근거다. 실제 광고 집행 이후의 매출, 구매 전환, 광고비, 도달 중복과 투자수익률을 직접 측정하거나 예측하지 않는다.</p>

- 온라인 관심과 감성은 실제 구매 의도나 광고 효과와 동일하지 않다.
- 높은 부정 반응은 논란뿐 아니라 극의 전개나 등장인물 맥락에서 발생할 수 있으므로 사람이 원인을 검토한다.
- 원천의 연령 구간, 시청 규모와 시청층 비율의 모집단, 키워드·감성 제공 범위를 실제 수집 후 확인한다.
- 데이터가 없거나 품질 기준을 통과하지 못하면 후보명이나 수치를 임의로 채우지 않는다.
- 최종 후보는 ‘집행 확정 매체’가 아니라 예산·계약·브랜드 세이프티를 포함한 후속 검토의 우선 대상이다.
